```python
import sys
from pathlib import Path

import pandas as pd
from openbb import obb

repo_root = Path.cwd().resolve()
if not (repo_root / "test_001_nvda").exists():
    repo_root = repo_root.parent
sys.path.insert(0, str(repo_root / "test_001_nvda"))

from config import build_model_path, load_strategy_config

config = load_strategy_config()
MODEL_PATH = build_model_path(repo_root, config["model_name"])

data = obb.equity.price.historical(
    config["stock_symbol"], provider=config["data_src"]
).to_df()

data["SMA20"] = data["close"].rolling(window=20).mean()
data["SMA50"] = data["close"].rolling(window=50).mean()
data[["close", "SMA20", "SMA50"]].plot(
    figsize=(10, 6), title=f"{config['stock_symbol']} Research"
 )
```

```python
import numpy as np
import pandas as pd
import xgboost as xgb
from openbb import obb

df = obb.equity.price.historical(
    config["stock_symbol"], provider=config["data_src"]
).to_df()

df["return"] = df["close"].pct_change()
df["vol_change"] = df["volume"].pct_change()
df["target"] = (df["return"].shift(-1) > 0).astype(int)

features = ["return", "vol_change"]
X = df[features].dropna()[:-1]
y = df["target"].loc[X.index]

model = xgb.XGBClassifier(
    n_estimators=10,
    max_depth=3,
    learning_rate=0.1,
 )
model.fit(X, y)

model.save_model(MODEL_PATH)
print(f"{config['model_name']}.json generated in {MODEL_PATH.parent}/")
```
